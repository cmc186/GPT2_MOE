# GENERADOR

```python
# Import modules
import random, math
import pandas as pd
import os, string

DATA_PATH = '../../../data'
```

#

## 1. MATH

```python
def generate_math_data(num_samples: int, values: tuple, save_path: str, seed=42, text: str = None, add: bool = True, result: bool = False) -> None:
    '''
    Generate math data for addition, subtraction, multiplication and division.
    Parameters:
    -----------
    - num_samples: number of samples to generate
    - values: tuple with the minimum and maximum numbers possible
    - save_path: path to save the generated data
    - seed: random seed
    - text: template for the prompt
    - add: if True, addition and subtraction (+, -); if False, multiplication and division (*, /)
    - result: if True, the solution is also written in the prompt ('{prompt} = {y}')
    '''
    # Define seed and range of values
    random.seed(seed)
    min_val, max_val = values
    rows = []

    # Base prompt
    if text is None:
        text = '{a} {operator} {b}'
    operators = ['+', '-'] if add else ['*', '/']

    # Generate samples and the truth predicction
    try:
        for _ in range(num_samples):
            operator = random.choice(operators)
            if operator == '/':
                b = random.choice([n for n in range(min_val, max_val + 1) if n != 0])
                a = b * random.randint(min_val, max_val)
                answer = a // b
            else:
                a, b = random.randint(min_val, max_val), random.randint(min_val, max_val)
                answer = a + b if operator == '+' else a - b if operator == '-' else a * b
            prompt, y = text.format(a=a, operator=operator, b=b), str(answer)
            rows.append([f'{prompt} = {y}' if result else prompt, y])
    except Exception as e:
        print(f'Error generating data: {e}')
        return

    # Save the generated data to a CSV file
    df = pd.DataFrame(rows, columns=['prompt', 'y']).astype(str)

    if os.path.exists(save_path):
        df.to_csv(save_path, mode='a', header=False, index=False)
    else:
        folder = os.path.dirname(save_path)
        if folder:
            os.makedirs(folder, exist_ok=True)
        df.to_csv(save_path, index=False)
```

## 2. PAIR

```python
def generate_parity_data(num_samples: int, values: tuple, save_path: str, seed=42, text: str = None, result: bool = False) -> None:
    '''
    Generate random numbers and their parity (even = 0 | odd = 1).
    Parameters:
    -----------
    - num_samples: number of samples to generate
    - values: tuple with the minimum and maximum numbers possible
    - save_path: path to save the generated data
    - seed: random seed
    - text: template for the prompt
    - result: if True, the solution is also written in the prompt ('{prompt} = {y}')
    '''
    # Define seed and range of values
    random.seed(seed)
    min_val, max_val = values
    rows = []

    # Base prompt
    if text is None:
        text = '{number}'

    # Generate samples and the truth predicction
    try:
        for _ in range(num_samples):
            n = random.randint(min_val, max_val)
            prompt, y = text.format(number=n), str(n % 2)
            rows.append([f'{prompt} = {y}' if result else prompt, y])
    except Exception as e:
        print(f'Error generating data: {e}')
        return

    # Save the generated data to a CSV file
    df = pd.DataFrame(rows, columns=['prompt', 'y']).astype(str)

    if os.path.exists(save_path):
        df.to_csv(save_path, mode='a', header=False, index=False)
    else:
        folder = os.path.dirname(save_path)
        if folder:
            os.makedirs(folder, exist_ok=True)
        df.to_csv(save_path, index=False)
```

## 3. ALPHABETIC ORDER

```python
def generate_alphabetic_data(num_samples: int, save_path: str, lenght: tuple, seed=42, text: str = None, result: bool = False) -> None:
    '''
    Generate random strings and their alphabetic order.
    Parameters:
    -----------
    - num_samples: number of samples to generate
    - save_path: path to save the generated data
    - lenght: tuple with the minimum and maximum length of the strings
    - seed: random seed
    - text: template for the prompt
    - result: if True, the solution is also written in the prompt ('{prompt} = {y}')
    '''
    # Define symbols, random seed and length range
    symbols = list(string.ascii_lowercase + string.ascii_uppercase)
    random.seed(seed)
    min_len, max_len = lenght
    rows = []

    # Base prompt
    if text is None:
        text = '{letters}'

    # Generate samples and the truth predicction
    try:
        for _ in range(num_samples):
            n = random.randint(min_len, max_len)
            letters = random.choices(symbols, k=n)
            ordered = sorted(letters, key=lambda c: (c.lower(), c.isupper()))
            prompt, y = text.format(letters=' '.join(letters)), ' '.join(ordered)
            rows.append([f'{prompt} = {y}' if result else prompt, y])
    except Exception as e:
        print(f'Error generating data: {e}')
        return

    # Save the generated data to a CSV file
    df = pd.DataFrame(rows, columns=['prompt', 'y']).astype(str)

    if os.path.exists(save_path):
        df.to_csv(save_path, mode='a', header=False, index=False)
    else:
        folder = os.path.dirname(save_path)
        if folder:
            os.makedirs(folder, exist_ok=True)
        df.to_csv(save_path, index=False)
```

## 4. SPELL

```python
def generate_spell_data(num_samples: int, save_path: str, lenght: tuple, seed=42, text: str = None, result: bool = False) -> None:
    '''
    Generate random strings and their spellings (space-separated).
    Parameters:
    -----------
    - num_samples: number of samples to generate
    - save_path: path to save the generated data
    - lenght: tuple with the minimum and maximum length of the strings
    - seed: random seed
    - text: template for the prompt
    - result: if True, the solution is also written in the prompt ('{prompt} = {y}')
    '''
    # Define symbols, random seed and length range
    symbols = list(string.ascii_lowercase) + list(string.ascii_uppercase) + list(string.digits) + list(string.punctuation)
    random.seed(seed)
    min_len, max_len = lenght
    rows = []

    # Base prompt
    if text is None:
        text = '{letters}'

    # Generate samples and the truth predicction
    try:
        for _ in range(num_samples):
            n = random.randint(min_len, max_len)
            letters = random.choices(symbols, k=n)
            word = ''.join(letters)
            prompt, y = text.format(letters=word), ' '.join(word)
            rows.append([f'{prompt} = {y}' if result else prompt, y])
    except Exception as e:
        print(f'Error generating data: {e}')
        return

    # Save the generated data to a CSV file
    df = pd.DataFrame(rows, columns=['prompt', 'y']).astype(str)

    if os.path.exists(save_path):
        df.to_csv(save_path, mode='a', header=False, index=False)
    else:
        folder = os.path.dirname(save_path)
        if folder:
            os.makedirs(folder, exist_ok=True)
        df.to_csv(save_path, index=False)
```

## 5. FIRST + LAST

```python
def generate_first_last_data(num_samples: int, save_path: str, lenght: tuple, seed=42, text: str = None, result: bool = False) -> None:
    '''
    Generate random strings of letters and return the first and last character concatenated.
    Parameters:
    -----------
    - num_samples: number of samples to generate
    - save_path: path to save the generated data
    - lenght: tuple with the minimum and maximum length of the strings
    - seed: random seed
    - text: template for the prompt
    - result: if True, the solution is also written in the prompt ('{prompt} = {y}')
    '''
    # Define symbols, random seed and length range
    symbols = list(string.ascii_lowercase + string.ascii_uppercase)
    random.seed(seed)
    min_len, max_len = lenght
    rows = []

    # Base prompt
    if text is None:
        text = '{letters}'

    # Generate samples and the truth predicction
    try:
        for _ in range(num_samples):
            n = random.randint(min_len, max_len)
            word = ''.join(random.choices(symbols, k=n))
            prompt, y = text.format(letters=word), word[0] + word[-1]
            rows.append([f'{prompt} = {y}' if result else prompt, y])
    except Exception as e:
        print(f'Error generating data: {e}')
        return

    # Save the generated data to a CSV file
    df = pd.DataFrame(rows, columns=['prompt', 'y']).astype(str)

    if os.path.exists(save_path):
        df.to_csv(save_path, mode='a', header=False, index=False)
    else:
        folder = os.path.dirname(save_path)
        if folder:
            os.makedirs(folder, exist_ok=True)
        df.to_csv(save_path, index=False)
```

```python
# Examples
def generate_all(num_samples: int = 10000, out_dir: str = 'data', seed: int = 42) -> None:
    generate_math_data(num_samples, (1, 10000), f'{out_dir}/add_sub.csv', seed=seed, add=True)
    generate_math_data(num_samples, (1, 10000), f'{out_dir}/mul_div.csv', seed=seed, add=False)
    generate_alphabetic_data(num_samples, f'{out_dir}/alphabetic.csv', lenght=(3, 25), seed=seed)
    generate_spell_data(num_samples, f'{out_dir}/spell.csv', lenght=(3, 25), seed=seed)
    generate_parity_data(num_samples, (0, 10000), f'{out_dir}/parity.csv', seed=seed)
    generate_first_last_data(num_samples, f'{out_dir}/first_last.csv', lenght=(3, 25), seed=seed)

generate_all(num_samples=25000, out_dir=DATA_PATH, seed=42)
```
